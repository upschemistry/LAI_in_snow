```python
import numpy as np
import matplotlib.pyplot as plt
```

The loadings below are uncorrected ISSW retrievals of the data in UPS_Rainier2025-07-03-01/, for

```python
ISSW_loadings = np.array([
32.491,
22.1128,
10.6280,
11.3788,
19.2907,
19.3581,
18.1347 ])
```

```python
labels = [
'1A.txt',
'1B.txt',
'2A.txt',
'3A.txt',
'4A.txt',
'5A.txt',
'6.txt']
```

These loadings are from the LAHM

```python
LAHM_loadings = np.array([
11.39,
9.7291,
8.276,
8.372,
8.0624,
8.8347,
6.5480])
```

```python
LAHM_loadings_extended = np.append(0,LAHM_loadings)
ISSW_loadings_extended = np.append(0,ISSW_loadings)
labels_extended = np.append('blank',labels); print(labels_extended)

plt.figure()
for i in range(len(LAHM_loadings_extended)):
    plt.scatter(LAHM_loadings_extended[i],ISSW_loadings_extended[i],label=labels_extended[i])
plt.xlabel(r'LAHM loading ($\mu g/filter$)')
plt.ylabel('ISSW loading at 450 nm (unscaled)')
plt.grid(True)
plt.legend()

weights = np.ones(np.shape(LAHM_loadings_extended))
weights[0]=100
p = np.polyfit(LAHM_loadings_extended,ISSW_loadings_extended,1,w=weights); print(p)
ISSW_loadings_extended_fit = np.polyval(p,LAHM_loadings_extended)
plt.plot(LAHM_loadings_extended,ISSW_loadings_extended_fit)

plt.figure()
for i in range(len(LAHM_loadings_extended)):
    plt.scatter(LAHM_loadings_extended[i],ISSW_loadings_extended[i]/p[0],label=labels_extended[i])
plt.xlabel(r'LAHM loading ($\mu g/filter$)')
plt.ylabel('ISSW loading at 450 nm (corrected)')
plt.grid(True)
plt.legend()
plt.plot(LAHM_loadings_extended,ISSW_loadings_extended_fit/p[0])
```
